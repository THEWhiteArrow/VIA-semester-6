```python
import seaborn as sns

arr = [
    3.775302,	3.350679,	4.217981,	4.030324,	4.639692,
4.139665	,4.395575,	4.824257	,4.268119	,4.584193,
4.930027	,4.315973,	4.600101,
]

mean = sum(arr) / len(arr)
std = (sum((x - mean) ** 2 for x in arr) / len(arr)) ** 0.5

standardized_arr = [(x - mean) / std for x in arr]
print("Original array:", arr)
print("Mean:", mean)
print("Standard Deviation:", std)
print("Standardized array:", standardized_arr)






import matplotlib.pyplot as plt

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

sns.histplot(arr, kde=True, ax=ax)
ax.set_title('Histogram with KDE')
ax.set_xlabel('Value')
ax.set_ylabel('Frequency')
plt.show()
 
import scipy.stats as stats
import numpy as np

# Create a probability plot
fig, ax = plt.subplots(figsize=(10, 6))
res = stats.probplot(arr, plot=ax)

# Add 95% confidence interval lines
slope, intercept, r = res[1]
x = np.linspace(min(arr), max(arr), 100)
y = slope * x + intercept

# Customize the plot
ax.set_title('Probability Plot with 95% Confidence Interval')
ax.legend()
plt.show()
```
